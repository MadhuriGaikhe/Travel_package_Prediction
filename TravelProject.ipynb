```python
df['DurationOfPitch'] = df['DurationOfPitch'].fillna(df['DurationOfPitch'].median())
df['NumberOfFollowups'] = df['NumberOfFollowups'].fillna(df['NumberOfFollowups'].median())
df['TypeofContact'] = df['TypeofContact'].fillna(df['TypeofContact'].mode().iloc[0])
df['PreferredPropertyStar'] = df['PreferredPropertyStar'].fillna(df['PreferredPropertyStar'].mean())
df['NumberOfTrips'] = df['NumberOfTrips'].fillna(df['NumberOfTrips'].mean())
df['NumberOfChildrenVisiting'] = df['NumberOfChildrenVisiting'].fillna(df['NumberOfChildrenVisiting'].mean())
df['MonthlyIncome'] = df['MonthlyIncome'].fillna(df['MonthlyIncome'].median())
```

# Model Training

```python
# separate features 
x = df.drop('ProdTaken',axis = 1)
y = df['ProdTaken']
```

```python
from sklearn.model_selection import train_test_split
```

```python
#Train test split
X_train,X_test,Y_tain,Y_test = train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
#PreProcessing
# Column Transformer = used to apply different Prepracessing technique on different column
```

```python
from sklearn.compose import ColumnTransformer
```

```python
cat_feature = x.select_dtypes(include='str').columns
num_feature = x.select_dtypes(exclude='str').columns
```

```python
cat_feature
```

```python
num_feature
```

```python
from sklearn.preprocessing import OneHotEncoder,StandardScaler
```

```python
#create an object of both preprocessing
```

```python
numeric_transformer = StandardScaler()
oh_transformer = OneHotEncoder(handle_unknown="ignore")
```

```python
preprocessor = ColumnTransformer(
    [
        ("OneHotEncoder",oh_transformer,cat_feature),
        ("StandardScaler",numeric_transformer,num_feature)
    ]
)
```

```python
# Appy Transformation on training Data 
```

```python
X_train = preprocessor.fit_transform(X_train)
X_test = preprocessor.transform(X_test)
```

```python
%pip install xgboost
```

```python
from sklearn.ensemble import RandomForestClassifier,AdaBoostClassifier,GradientBoostingClassifier
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
```

```python
# create boject of all models
models = {
    "Random Forest":RandomForestClassifier(),
    "Ada boost" : AdaBoostClassifier(),
    "Gradient boost": GradientBoostingClassifier(),
    "Decision tree" : DecisionTreeClassifier(),
    "Xgboosst" : XGBClassifier()
}
```

```python
for i in range(len(list(models))):
    model = list(models.values())[i]

    # model Training
    model.fit(X_train,Y_tain)

    # make prediction
    y_pred_train = model.predict(X_train)
    y_pred_test = model.predict(X_test)

    print(list(models.keys())[i])
    print("Training Accuracy")
    print(f"accuracy_score",accuracy_score(Y_tain,y_pred_train))
    print(f"precision_score",precision_score(Y_tain,y_pred_train))
    print(f"recall_score",recall_score(Y_tain,y_pred_train))
    print(f"f1_score",f1_score(Y_tain,y_pred_train))

    print("----------------------------------------------------------------")
    
    print("Testing Accuracy")
    print(f"accuracy_score",accuracy_score(Y_test,y_pred_test))
    print(f"precision_score",precision_score(Y_test,y_pred_test))
    print(f"recall_score",recall_score(Y_test,y_pred_test))
    print(f"f1_score",f1_score(Y_test,y_pred_test))
    print("_______________________________________________________________")

```
